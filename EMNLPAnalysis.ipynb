```python
import json
import os
from collections import defaultdict

import pandas as pd

experiments = set()
restr_directories = ["main_logits", "original_logits", "root_logits"]

uas_pd_results = defaultdict(dict)
las_pd_results = defaultdict(dict)

data_dict = {}

results = defaultdict(dict)
for restr_dir in restr_directories:
    for filename in os.listdir(restr_dir):
        if "metrics" in filename:
            with open(f"{restr_dir}/{filename}", 'r', encoding='utf-8') as file:
                exp_name, restr_name = filename.replace("metrics_", "").replace(".jsonl", "").split("_42_")
                assert restr_dir in restr_name
                data = json.load(file)
                filename_prefix = filename.split(".")[0]
                if filename_prefix + "_uas" in data:
                    uas_pd_results[exp_name][restr_dir] = data[filename_prefix + "_uas"]
                if filename_prefix + "_las" in data:
                    las_pd_results[exp_name][restr_dir] = data[filename.split(".")[0] + "_las"]
                experiments.add(exp_name)
                data_dict[filename] = data

wrong_experiments = {"Qwen3_4B_en_ewt_grct_en_ewt_grct", "Qwen3_0_6B_en_ewt_grct_en_ewt_grct",
                "Qwen3_8B_en_ewt_grct_en_ewt_grct", "Qwen3_8B_fr_gsd_fr_gsd", "Qwen3_0_6B_fr_gsd_fr_gsd"}
for exp in experiments:
  if exp not in wrong_experiments: # TODO
    print(exp)
    uas_pd_results[exp] = pd.DataFrame(uas_pd_results[exp])
    las_pd_results[exp] = pd.DataFrame(las_pd_results[exp])
```
